```python
# 1. Generate a vector x (representing a 1D signal) with a size of 10, initialized to all zeros.

x = np.zeros(10)

print("Vector x:")
print(x)

print("Shape of x:", x.shape)
```

```python
# 2. Generate a matrix A (representing a 2D signal) with random values in the range of [0, 255] and a shape of (16,16).

A = np.random.randint(0, 256, (16,16))

print("Matrix A:")
print(A)

print("Shape of A:", A.shape)
```

```python
# 3. Create matrix B by flipping A horizontally (mirroring the columns).

B = np.fliplr(A)

print("Matrix B (flipped horizontally):")
print(B)

print("Shape of B:", B.shape)
```

```python
# 4. Change the 4th element of vector x to a value of 0.

x[3] = 0

print("Vector x after changing the 4th element:")
print(x)
```

```python
# 5. Delete the 4th element of vector x. Observe how the signal length shrinks.

x = np.delete(x, 3)

print("Vector x after deleting the 4th element:")
print(x)

print("New length of x:", len(x))
print("New shape of x:", x.shape)
```

```python
# 6. Replace the odd rows from matrix A by the matrix B.

A[1::2] = B[1::2]

print("Matrix A after replacing odd rows:")
print(A)

print("Shape of A:", A.shape)
```

```python
# 7. Delete the even columns from matrix A.

A = np.delete(A, np.s_[1::2], axis=1)

print("Matrix A after deleting even columns:")
print(A)

print("New shape of A:", A.shape)
```

```python
# 8. Calculate the sum between matrices A and B, and normalize the output values to the range of [0,255].

B_resized = np.delete(B, np.s_[1::2], axis=1)

C = A + B_resized
C_normalized = ((C - C.min()) / (C.max() - C.min()) * 255).astype(np.uint8)

print("Summed and normalized matrix:")
print(C_normalized)
```

```python
# Visualization: Display the normalized matrix as an image

plt.imshow(C_normalized, cmap='gray')
plt.title("Normalized Matrix")
plt.colorbar()
plt.show()
```
